```python
ip=dataset.drop(["Heating_Load","Cooling_Load","Orientation"],axis=1)
ip
```

```python
op=dataset[["Heating_Load","Cooling_Load"]]
op
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(ip,op,train_size=0.8)
```

```python
xtrain
```

```python
xtest
```

```python
ytrain
```

```python
ytest
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
```

```python
xtrain=sc.fit_transform(xtrain)
xtrain
```

```python
xtest=sc.transform(xtest)
xtest
```

# Linear regression

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(xtrain,ytrain)
```

```python
pred=lr.predict(xtest)
pred
```

```python
ytest
```

```python
from sklearn.metrics import r2_score
r2=r2_score(ytest,pred)
r2
```

# polynomial regression

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(xtrain)
pol_reg=LinearRegression()

pol_reg.fit(x_poly,ytrain)
x1=poly.fit_transform(xtest)
yp2=pol_reg.predict(x1)
```

```python
# accuracy
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(ytest,yp2)
r2=r2_score(ytest,yp2)
print(mse)
print(r2)
```

# lasso

```python
from sklearn.linear_model import Lasso
L=Lasso(alpha=0.01)
L.fit(xtrain,ytrain)
yp3=L.predict(xtest)
```

```python
# accuracy
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(ytest,yp3)
r2=r2_score(ytest,yp3)
print(mse)
print(r2)
```

# Ridge

```python
from sklearn.linear_model import Ridge
r=Ridge(alpha=1)
r.fit(xtrain,ytrain)
yp4=r.predict(xtest)
```

```python
# accuracy
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(ytest,yp4)
r2=r2_score(ytest,yp4)
print(mse)
print(r2)
```
